```python
# Setup
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from data import load_data, RAW_DIR
from utils import (save_fig, plot_class_distribution,
                   plot_review_length_distribution, plot_top_words_per_class)

pd.set_option('display.max_colwidth', 200)
sns.set_style('whitegrid')
print(f"Raw data dir: {RAW_DIR}")
```

```python
# 1. Load data
df = load_data()
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
df.head()
```

```python
# 2. Class distribution
plot_class_distribution(df)
save_fig('01_class_distribution.png')
plt.show()
print(df['sentiment'].value_counts())
```

```python
# 3. Review length distribution by sentiment
plot_review_length_distribution(df)
save_fig('02_length_distribution.png')
plt.show()

df['length'] = df['text'].str.len()
print(df.groupby('sentiment')['length'].describe().round(0))
```

```python
# 4. Top words per class
axes = plot_top_words_per_class(df, top_n=15)
save_fig('03_top_words_per_class.png')
plt.show()
```

```python
# 5. Sample reviews per class
for cls in ['positive', 'neutral', 'negative']:
    print(f"\n=== {cls.upper()} ===")
    for r in df[df['sentiment'] == cls]['text'].head(3):
        print(f"  - {r}")
```
